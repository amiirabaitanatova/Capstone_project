# Описание и подготовка данных

# Анализ и визуализация данных

```python
score_counts = df['review_score'].value_counts(normalize=True).sort_index() * 100
colors = ['#d73027' if val < 10 else '#fee08b' if val < 30 else '#1a9850' for val in score_counts.values]

fig, ax = plt.subplots(figsize=(9, 6))
bars = ax.bar(score_counts.index, score_counts.values, color=colors, width=0.6)

labels = [f'{val:.1f}%' for val in score_counts.values]
ax.bar_label(bars, labels=labels, padding=3, fontsize=11, fontweight='bold')

ax.set_title('Распределение оценок клиентов', fontsize=13, pad=20)
ax.set_xlabel('Оценка (звёзды)', fontsize=11)
ax.set_ylabel('Доля заказов, %', fontsize=11)
sns.despine(left=True)
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

* Почти 80% всех покупателей ставят отличные оценки — 4 или 5 звёзд. Чаще всего ставят ровно 5 звёзд, около 60% всех заказов.
* Двоек и троек очень мало. Покупатели обычно ставят либо «отлично» если всё прошло гладко, либо сразу «1 звезду» если что-то сильно испортило впечатление.

# Модели машинного обучения

```python
# Бинарная классификация целевой переменной
df['target'] = (df['review_score'] >= 4).astype(int)
df['target'].value_counts(normalize=True)
```

```python
df.drop(columns=['review_score'], inplace=True)
```

```python
y = df['target']
X = df.drop(columns=['target'])
```

```python
# Разделение на train/test 
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42, stratify=y)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler

num_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()

# Для числовых признаков заполняем пропуски медианой
num_imputer = SimpleImputer(strategy='median')
X_train_num_imp = num_imputer.fit_transform(X_train[num_cols])
X_test_num_imp = num_imputer.transform(X_test[num_cols])

# Для категориальных признаков заполняем пропуски самым частым значением или 'Unknown'
cat_imputer = SimpleImputer(strategy='constant', fill_value='Unknown')
X_train_cat_imp = cat_imputer.fit_transform(X_train[cat_cols])
X_test_cat_imp = cat_imputer.transform(X_test[cat_cols])

# Масштабируем числовые признаки
scaler = StandardScaler()
X_train_num_scaled = scaler.fit_transform(X_train_num_imp)
X_test_num_scaled = scaler.transform(X_test_num_imp)

# Кодируем категориальные признаки
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
X_train_cat_ohe = ohe.fit_transform(X_train_cat_imp)
X_test_cat_ohe = ohe.transform(X_test_cat_imp)

# Объединяем обработанные признаки обратно
X_train_final = np.hstack([X_train_num_scaled, X_train_cat_ohe])
X_test_final = np.hstack([X_test_num_scaled, X_test_cat_ohe])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score, f1_score

log_reg = LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42)
log_reg.fit(X_train_final, y_train)

y_pred_log_reg = log_reg.predict(X_test_final)
y_proba_log_reg = log_reg.predict_proba(X_test_final)[:, 1]

print(f'ROC-AUC: {roc_auc_score(y_test, y_proba_log_reg):.3f}')
print(classification_report(y_test, y_pred_log_reg))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, max_depth=10, class_weight='balanced', random_state=42)
rf.fit(X_train_final, y_train)

y_pred_rf = rf.predict(X_test_final)
y_proba_rf = rf.predict_proba(X_test_final)[:, 1]

print(f"ROC-AUC: {roc_auc_score(y_test, y_proba_rf):.3f}")
print(classification_report(y_test, y_pred_rf))
```

```python
from catboost import CatBoostClassifier

cb = CatBoostClassifier(auto_class_weights='Balanced', verbose=0, random_state=42)
cb.fit(X_train_final, y_train)

y_pred_cb = cb.predict(X_test_final)
y_proba_cb = cb.predict_proba(X_test_final)[:, 1]

print(f"ROC-AUC: {roc_auc_score(y_test, y_proba_cb):.3f}")
print(classification_report(y_test, y_pred_cb))
```

```python
from sklearn.metrics import roc_curve, auc

# Logistic Regression
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_proba_log_reg)
auc_lr = auc(fpr_lr, tpr_lr)

# Random Forest
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_proba_rf)
auc_rf = auc(fpr_rf, tpr_rf)

# CatBoost
fpr_cb, tpr_cb, _ = roc_curve(y_test, y_proba_cb)
auc_cb = auc(fpr_cb, tpr_cb)

plt.figure(figsize=(9, 6))

# ROC-кривые для каждой модели
plt.plot(fpr_lr, tpr_lr, label=f'Logistic Regression (AUC = {auc_lr:.2f})', color='#3b62a6', linewidth=2)
plt.plot(fpr_rf, tpr_rf, label=f'Random Forest (AUC = {auc_rf:.2f})', color='#4ca862', linewidth=2)
plt.plot(fpr_cb, tpr_cb, label=f'CatBoost (AUC = {auc_cb:.2f})', color='#c84b4b', linewidth=2)

# Диагональная линия случайного угадывания (Random Classifier)
plt.plot([0, 1], [0, 1], color='gray', linestyle='--', linewidth=1.5)

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate', fontsize=11)
plt.ylabel('True Positive Rate', fontsize=11)
plt.title('ROC-кривая для моделей бинарной классификации', fontsize=13, fontweight='bold', pad=15)

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='lower right', fontsize=10, frameon=True)

plt.tight_layout()
plt.show()
```

```python
# Список всех имен признаков после OneHotEncoder и Imputer
cat_feature_names = ohe.get_feature_names_out(cat_cols).tolist()
all_features = np.array(num_cols + cat_feature_names)

# Извлекаем важности признаков для каждой модели

lr_importance = np.abs(log_reg.coef_[0])
rf_importance = rf.feature_importances_
cb_importance = cb.feature_importances_

fi_df = pd.DataFrame({
    'Feature': all_features,
    'Logistic Regression': lr_importance / lr_importance.sum(), # Нормируем к 1 
    'Random Forest': rf_importance,
    'CatBoost': cb_importance / cb_importance.sum()
})

# Топ-10 самых важных признаков по версии CatBoost
top10_features = fi_df.sort_values(by='CatBoost', ascending=False).head(10)
top10_features = top10_features.set_index('Feature')
top10_features
```

```python
ax = top10_features.plot(kind='barh', figsize=(12, 7), color=['#3b62a6', '#4ca862', '#c84b4b'], width=0.8)

plt.title('Сравнение важности топ 10 признаков для трех моделей', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Относительная важность признака', fontsize=12)
plt.ylabel('Признак', fontsize=12)
plt.gca().invert_yaxis() 
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.legend(title='Модель', fontsize=10)
plt.tight_layout()
plt.show()
```

#### Вывод:

* Главный источник негатива — это задержка заказа (delivery_delay_days) и общее время в пути (total_delivery_days). Если посылка едет долго или опаздывает к обещанному сроку, клиент почти наверняка поставит низкую оценку. Это подтвердили абсолютно все алгоритмы.
